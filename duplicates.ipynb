1. Print CSV
2. Identify fully duplicated rows
3. Remove exact duplicates
4. Check for duplicates based on [order_id] only

```python
import pandas as pd

df = pd.read_csv('orders.csv')
print(df.to_string())
```

```python
df.duplicated()
```

```python
df.drop_duplicates()
```

```python
df.duplicated(subset=['order_id'])
```

```python
no_duplicates = df.drop_duplicates()
print(f'Final row count: {len(no_duplicates)}')
```
